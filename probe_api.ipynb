```python
from dotenv import load_dotenv
load_dotenv()

from src.api.visitkorea_client import VisitKoreaClient
import json

# JpnService2 contentTypeId (KorService2와 다름)
CTYPE_ATTRACTION = "76"   # 관광지·명소
CTYPE_STAY       = "80"   # 숙박
CTYPE_FOOD       = "82"   # 음식점·카페
CTYPE_FESTIVAL   = "85"   # 축제·행사

BUSAN_AREA_CODE = "6"

client = VisitKoreaClient()

def fetch_all(area_code: str, content_type_id: str) -> list:
    items, page = [], 1
    while True:
        batch, _, _, total = client.search_attractions(
            area_code=area_code,
            content_type_id=content_type_id,
            page_no=page,
            num_of_rows=100,
        )
        items.extend(i.to_dict() for i in batch)
        if len(items) >= total or not batch:
            break
        page += 1
    return items

festivals   = fetch_all(BUSAN_AREA_CODE, CTYPE_FESTIVAL)    # 7건
attractions = fetch_all(BUSAN_AREA_CODE, CTYPE_ATTRACTION)  # 133건
stays       = fetch_all(BUSAN_AREA_CODE, CTYPE_STAY)        # 27건
cultures    = fetch_all(BUSAN_AREA_CODE, CTYPE_FOOD)        # 10건

result = {
    "area": "부산 (areaCode=6)",
    "festival":   {"total": len(festivals),   "items": festivals},
    "attraction": {"total": len(attractions), "items": attractions},
    "stay":       {"total": len(stays),       "items": stays},
    "culture":    {"total": len(cultures),    "items": cultures},
}

print(json.dumps(result, ensure_ascii=False, indent=2))
```

```python
import pandas as pd

for category, label in [
    ("festival",   "축제"),
    ("attraction", "관광지"),
    ("stay",       "숙박"),
    ("culture",    "음식점/카페"),
]:
    items = result[category]["items"]
    if not items:
        print(f"[{label}] 데이터 없음\n")
        continue
    df = pd.DataFrame(items)[["title", "addr1", "tel", "event_period", "maps_uri"]]
    df.columns = ["이름", "주소", "전화", "기간", "지도"]
    df = df.replace("", pd.NA)
    print(f"=== {label} ({len(items)}건) ===")
    display(df)
```

```python
# 전 지역 관광지(contentTypeId=76) 전수 조회
import pandas as pd

AREA_CODES = {
    "서울": "1", "인천": "2", "대전": "3", "대구": "4",
    "광주": "5", "부산": "6", "울산": "7", "세종": "8",
    "경기": "31", "강원": "32", "충북": "33", "충남": "34",
    "경북": "35", "경남": "36", "전북": "37", "전남": "38", "제주": "39",
}

all_rows = []
for region, code in AREA_CODES.items():
    items = fetch_all(code, "76")
    for item in items:
        item["region"] = region
        all_rows.append(item)
    print(f"{region}({code}): {len(items)}건")

df_all = pd.DataFrame(all_rows)[["region", "title", "addr1", "cat1", "cat2", "cat3", "maps_uri"]]
df_all.columns = ["지역", "이름", "주소", "cat1", "cat2", "cat3", "지도"]
df_all = df_all.replace("", pd.NA)
print(f"\n전체 합계: {len(df_all)}건")
display(df_all)
```
